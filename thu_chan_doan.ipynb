# Thử chẩn đoán lỗi và hỏi admin

Chạy ô 1 → 2 → 3. Ở ô 2, một câu hỏi y/n hiện ở **thanh trên cùng màn hình**. Thử trả lời `y`, rồi chạy lại ô 2 và trả lời `n` để xem khác nhau thế nào.

```python
# ===== 1. Tình huống cố ý làm khó: chỉ có 3 nhân viên =====
NHAN_VIEN = ["An", "Binh", "Chi"]
NGAY = ["Thứ 2", "Thứ 3", "Thứ 4"]
CA = ["Sáng", "Chiều", "Tối"]
SO_NGUOI = {"Sáng": 2, "Chiều": 2, "Tối": 2}

# Luật cứng: ai làm ca Sáng thì hôm đó không được làm ca Tối (và ngược lại)
CAP_CAM = [("Sáng", "Tối")]
# Mỗi ngày cần 2 người ca Sáng + 2 người ca Tối = 4 người KHÁC NHAU, mà chỉ có 3 nhân viên -> không xếp được
```

```python
# ===== 2. Máy thử xếp, nếu thất bại thì giải thích và hỏi admin =====
from solver import xep_lich_tu_van, lich_theo_ca

def hoi_admin(van_de):
    """Hàm này được gọi khi không xếp được mà nới luật thì xếp được. Trả về True nếu admin đồng ý."""
    print("KHÔNG XẾP ĐƯỢC LỊCH. Các vấn đề tìm thấy:")
    for v in van_de:
        print(" -", v["noi_dung"])
    print()
    tra_loi = input("Cho phép một người làm hai ca trong một ngày để xếp tiếp? (y/n): ")
    return tra_loi.strip().lower() in ("y", "yes", "co", "có")

ket = xep_lich_tu_van(NHAN_VIEN, NGAY, CA, SO_NGUOI, hoi=hoi_admin, cap_ca_cam=CAP_CAM)
```

```python
# ===== 3. Xem kết quả =====
kq = ket["ket_qua"]
print("Trạng thái:", kq["trang_thai"])

if kq["lich"] is None:
    print("Chưa có lịch. Các vấn đề cần xử lý:")
    for v in ket["van_de"]:
        print(" -", v["noi_dung"])
else:
    bang = lich_theo_ca(kq["lich"], NHAN_VIEN, NGAY, CA)
    for (ngay, ca), ds in bang.items():
        print(ngay, "-", ca, ":", ds)
    if ket["da_noi_long"]:
        print()
        print("Lịch này có người làm hai ca trong một ngày (do admin đồng ý):")
        for dong in ket["nguoi_lam_hai_ca"]:
            print(" -", dong)
```
