# Thử nhiều tuần: đăng ký → ghim → xếp → lưu → thống kê
Chạy từ trên xuống. File dữ liệu demo: `demo.db` (xóa đi để chạy lại từ đầu).

```
import os, database as db, solver as sv
if os.path.exists("demo.db"): os.remove("demo.db")
conn = db.ket_noi("demo.db")
for ten in ["An","Binh","Chi","Dung","Em","Phuc"]:
    db.them_nhan_vien(conn, ten)
db.them_nhan_vien(conn, "Giang", loai="full_time")   # full-time: admin tự ghim ca
db.dat_toi_da_tuan(conn, 7)                           # giới hạn ca/tuần (tùy chọn, đặt None để bỏ)
print(db.ds_nhan_vien_chi_tiet(conn))
```

## Mỗi tuần: nhân viên đăng ký, admin ghim ca cho full-time, rồi xếp

```
TUAN = ["2026-10-05", "2026-10-12", "2026-10-19"]
cfg0 = db.doc_cau_hinh(conn, TUAN[0])
tat_ca = {(d, c) for d in cfg0["ngay"] for c in cfg0["ca"]}
for tuan in TUAN:
    for ten in ["An","Binh","Chi","Dung","Em","Phuc"]:
        db.luu_dang_ky(conn, tuan, ten, tat_ca)                       # ở đây cho ai cũng làm được hết
    db.dat_ghim(conn, tuan, "Giang", {("Thứ 2","Sáng"), ("Thứ 3","Sáng")})
    cfg = db.doc_cau_hinh(conn, tuan)
    r = sv.xep_lich(**cfg)
    print(tuan, r["trang_thai"], "số ca từng người:", r["so_ca"])
    db.luu_lich(conn, tuan, sv.lich_theo_ca(r["lich"], cfg["nhan_vien"], cfg["ngay"], cfg["ca"]))
```
