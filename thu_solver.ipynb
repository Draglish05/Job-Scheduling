# Thử `solver.py` với nhà hàng 12 nhân viên

Notebook này không còn `input()`: dữ liệu khai báo ở ô 1, ô 2 gọi hàm `xep_lich`.

```python
# ===== 1. Cấu hình nhà hàng và dữ liệu đăng ký =====
import random
from solver import xep_lich, lich_theo_ca, kiem_tra_du_lieu, kiem_tra_lich, canh_bao_lien_nhau, canh_bao_cung_ngay

NHAN_VIEN = ["An", "Binh", "Chi", "Dung", "Em", "Phuc", "Giang", "Hoa", "Khoa", "Lan", "Minh", "Nga"]
NGAY = ["Thứ 2", "Thứ 3", "Thứ 4", "Thứ 5", "Thứ 6", "Thứ 7", "Chủ nhật"]
CA = ["Sáng", "Chiều", "Tối"]                       # nhà hàng khác có thể đổi: ["Sáng", "Tối"]...
SO_NGUOI = {"Sáng": 2, "Chiều": 2, "Tối": 3}        # số người mỗi loại ca

# Dữ liệu đăng ký GIẢ để thử: mỗi người đăng ký ngẫu nhiên khoảng 60% số ca
random.seed(3)
LAM_DUOC = {
    ten: {(d, c) for d in NGAY for c in CA if random.random() < 0.6}
    for ten in NHAN_VIEN
}

print("Cảnh báo dữ liệu:", kiem_tra_du_lieu(NHAN_VIEN, NGAY, CA, SO_NGUOI, LAM_DUOC) or "không có")
```

```python
# ===== 2. Xếp lịch =====
CAP_HAN_CHE = [("Sáng", "Tối")]     # luật mềm: máy cố tránh việc ai làm ca Sáng thì hôm đó còn làm ca Tối
kq = xep_lich(NHAN_VIEN, NGAY, CA, SO_NGUOI, LAM_DUOC, toi_da_tuan=8, cap_ca_han_che=CAP_HAN_CHE)
print("Trạng thái:", kq["trang_thai"])

if kq["lich"] is not None:
    bang = lich_theo_ca(kq["lich"], NHAN_VIEN, NGAY, CA)
    for d in NGAY:
        for c in CA:
            print(d, "-", c, ":", bang[(d, c)])

    print()
    print("Số ca mỗi người:", dict(zip(NHAN_VIEN, kq["so_ca"])))
    print("Số cặp ca liền nhau trong cùng ngày:", kq["so_cap_lien_nhau"])
    print("Số lần làm Tối hôm trước rồi Sáng hôm sau:", kq["so_cap_qua_dem"])
    print("Lỗi luật cứng:", kiem_tra_lich(kq["lich"], NHAN_VIEN, NGAY, CA, SO_NGUOI, LAM_DUOC, toi_da_tuan=8) or "không có")
```
